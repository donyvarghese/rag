```python
### Text Loader

from langchain_community.document_loaders import TextLoader

loader = TextLoader("../data/text_files/python_intro.txt", encoding="utf-8")

loader

document = loader.load()

print(document)
```

```python
## Directory Loader

from langchain_community.document_loaders import DirectoryLoader

## Load all text files in a directory

dir_loader = DirectoryLoader("../data/text_files", glob="*.txt", loader_cls=TextLoader, loader_kwargs={"encoding": "utf-8"})

documents = dir_loader.load()
documents
```

```python
# PDF Loader

from langchain_community.document_loaders import PyPDFLoader, PyMuPDFLoader, DirectoryLoader

dir_loader = DirectoryLoader("../data/pdf", glob="*.pdf", loader_cls=PyMuPDFLoader)

pdf_documents = dir_loader.load()
pdf_documents
```

```python
# CSV Loader

from langchain_community.document_loaders import CSVLoader, DirectoryLoader

csv_loader = DirectoryLoader("../data/csv", glob="*.csv", loader_cls=CSVLoader)

csv_documents = csv_loader.load()
csv_documents
```
